```python
%sql
--    One External Volume per source. Volume name matches the source identifier
--    used in the watermark and in the Silver table name, so each Silver
--    notebook reads from /Volumes/uk_property_intel/bronze/<source>/ and
--    writes to uk_property_intel.silver.<source>.

CREATE EXTERNAL VOLUME IF NOT EXISTS uk_property_intel.bronze.boe
  LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/boe/base_rate/'
  COMMENT 'Bank of England rate files. Currently contains base_rate/; structure anticipates additional rate types in future.';

CREATE EXTERNAL VOLUME IF NOT EXISTS uk_property_intel.bronze.hpi
  LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/uk_hpi/'
  COMMENT 'HM Land Registry UK House Price Index, cumulative monthly CSV.';

CREATE EXTERNAL VOLUME IF NOT EXISTS uk_property_intel.bronze.ppd
  LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/land_registry/ppd/'
  COMMENT 'HM Land Registry Price Paid Data, yearly CSVs from 1995.';

DROP VOLUME IF EXISTS uk_property_intel.bronze.doogal;

CREATE EXTERNAL VOLUME IF NOT EXISTS uk_property_intel.bronze.doogal
  LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/doogal/'
  COMMENT 'Doogal UK Postcode Lookup, ONSPD mirror, quarterly ZIP refresh.';

CREATE EXTERNAL VOLUME IF NOT EXISTS uk_property_intel.bronze.ons
  LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/ons/private_rent_index/'
  COMMENT 'ONS Price Index of Private Rents, monthly XLSX.';

CREATE EXTERNAL VOLUME IF NOT EXISTS uk_property_intel.bronze.police
  LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/police/crime/'
  COMMENT 'UK Police street-level crime, rolling 3-year snapshot ZIPs, 2-year-stepped historical + monthly latest.';
```
