# Create schemas for the UC. 
## - (bronze, silver, gold, quality)

```python
%sql
CREATE SCHEMA IF NOT EXISTS uk_property_intel.bronze
  MANAGED LOCATION 'abfss://bronze@ukpropertyintelligencedl.dfs.core.windows.net/'
  COMMENT 'Raw files as landed by ADF. External Volumes provide UC governance, lineage, and discoverability over the bronze container without converting raw data to Delta.';

CREATE SCHEMA IF NOT EXISTS uk_property_intel.silver
  MANAGED LOCATION 'abfss://silver@ukpropertyintelligencedl.dfs.core.windows.net/'
  COMMENT 'Cleaned, typed, deduped per-source tables. One table per Bronze source.';

CREATE SCHEMA IF NOT EXISTS uk_property_intel.gold
  MANAGED LOCATION 'abfss://gold@ukpropertyintelligencedl.dfs.core.windows.net/'
  COMMENT 'Multi-source enriched marts. Kimball-style facts and dimensions for analytical consumption.';

CREATE SCHEMA IF NOT EXISTS uk_property_intel.quality
  MANAGED LOCATION 'abfss://quality@ukpropertyintelligencedl.dfs.core.windows.net/'
  COMMENT 'Data quality framework outputs: quarantine tables for rejected records, rule run history, DQ metrics.';
```

```python
%sql
SHOW SCHEMAS IN uk_property_intel;
```

```python
%sql
DESCRIBE SCHEMA EXTENDED uk_property_intel.silver;
DESCRIBE SCHEMA EXTENDED uk_property_intel.gold;
DESCRIBE SCHEMA EXTENDED uk_property_intel.quality;
```
